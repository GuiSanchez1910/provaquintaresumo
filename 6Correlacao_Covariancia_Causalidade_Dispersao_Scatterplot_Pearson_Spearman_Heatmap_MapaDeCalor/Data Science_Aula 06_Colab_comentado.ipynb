```python
# [CÉLULA 1] SETUP DO AMBIENTE E BIBLIOTECAS
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Configurando o visual dos gráficos para apresentação
sns.set_theme(style="whitegrid")
```

```python
# [CÉLULA 2] SIMULAÇÃO: LENDO A NUVEM DE PONTOS

# Define uma "semente" para a geração de números aleatórios.
# Com seed(42), toda vez que o código rodar, os mesmos números
# aleatórios serão gerados.
# Isso faz com que o gráfico seja sempre igual.
np.random.seed(42)


# Cria 100 valores igualmente espaçados entre 0 e 100.
# Exemplo: 0, 1.01, 2.02, 3.03, ..., 100
# Esses valores serão usados como os valores de X.
x_base = np.linspace(0, 100, 100)


# ============================================================
# 1. TENDÊNCIA POSITIVA
# ============================================================

# Cria os valores de Y.
#
# x_base → existe uma relação positiva entre X e Y:
#           conforme X aumenta, Y tende a aumentar.
#
# np.random.normal(0, 10, 100) → adiciona "ruído" aleatório:
#   0  = média do ruído
#   10 = desvio padrão
#   100 = quantidade de valores gerados
#
# Portanto:
# Y = X + ruído
#
# Os pontos não ficam perfeitamente alinhados porque existe
# uma variação aleatória.
y_positivo = x_base + np.random.normal(0, 10, 100)


# ============================================================
# 2. TENDÊNCIA NEGATIVA
# ============================================================

# Cria uma relação negativa entre X e Y.
#
# 100 - x_base → conforme X aumenta, Y tende a diminuir.
#
# Novamente adicionamos um ruído aleatório para deixar
# os pontos mais realistas.
#
# Portanto:
# Y = 100 - X + ruído
y_negativo = 100 - x_base + np.random.normal(0, 10, 100)


# ============================================================
# 3. CAOS TOTAL / SEM CORRELAÇÃO
# ============================================================

# Cria 100 valores de X aleatórios.
#
# Média = 50
# Desvio padrão = 20
# Quantidade = 100
#
# Não existe uma relação definida entre X e Y.
x_caos = np.random.normal(50, 20, 100)


# Cria 100 valores de Y aleatórios.
#
# Também possui média 50 e desvio padrão 20.
#
# Como X e Y são gerados independentemente,
# não existe uma tendência clara entre eles.
y_caos = np.random.normal(50, 20, 100)


# ============================================================
# CRIANDO OS GRÁFICOS
# ============================================================

# Cria uma figura contendo 3 gráficos lado a lado.
#
# plt.subplots(1, 3):
#   1 = quantidade de linhas
#   3 = quantidade de colunas
#
# figsize=(18, 5):
#   largura = 18
#   altura = 5
#
# fig = figura completa
# axs = lista contendo os 3 gráficos
fig, axs = plt.subplots(1, 3, figsize=(18, 5))


# ============================================================
# GRÁFICO 1 — CORRELAÇÃO POSITIVA
# ============================================================

# Cria um gráfico de dispersão (scatter plot).
#
# x=x_base → valores do eixo X
# y=y_positivo → valores do eixo Y
# ax=axs[0] → coloca no primeiro gráfico
# color='blue' → define a cor azul
#
# .set_title() → define o título do gráfico
sns.scatterplot(
    x=x_base,
    y=y_positivo,
    ax=axs[0],
    color='blue'
).set_title('Correlação Positiva (+)')


# ============================================================
# GRÁFICO 2 — CORRELAÇÃO NEGATIVA
# ============================================================

# Cria o segundo gráfico de dispersão.
#
# Conforme X aumenta, Y tende a diminuir.
#
# ax=axs[1] → coloca no segundo gráfico
sns.scatterplot(
    x=x_base,
    y=y_negativo,
    ax=axs[1],
    color='red'
).set_title('Correlação Negativa (-)')


# ============================================================
# GRÁFICO 3 — SEM CORRELAÇÃO
# ============================================================

# Cria o terceiro gráfico de dispersão.
#
# X e Y foram gerados aleatoriamente e de forma independente.
# Por isso, não existe uma tendência clara.
#
# ax=axs[2] → coloca no terceiro gráfico
sns.scatterplot(
    x=x_caos,
    y=y_caos,
    ax=axs[2],
    color='gray'
).set_title('Sem Correlação (Caos)')


# Mostra os três gráficos na tela.
plt.show()
```

### [CÉLULA 3] O Ponto Cego de Pearson
O Coeficiente de Pearson avalia a força baseando-se em linhas retas. Quando o crescimento ocorre em curvas exponenciais, utilizamos o Ranking de Spearman.

**Fórmula de Pearson:**
$$ r = \frac{Cov(X,Y)}{s_x \cdot s_y} $$

```python
# [CÉLULA 4] O DUELO: PEARSON VS SPEARMAN

# ============================================================
# CRIANDO OS DADOS
# ============================================================

# Cria números inteiros de 1 até 19.
#
# np.arange(1, 20)
# → começa em 1
# → vai até antes do 20
#
# Resultado:
# [1, 2, 3, ..., 19]
x_curva = np.arange(1, 20)


# Cria Y elevando cada valor de X ao cubo.
#
# Exemplo:
# 1³ = 1
# 2³ = 8
# 3³ = 27
# 4³ = 64
#
# Isso cria uma relação CURVA entre X e Y.
y_curva = x_curva ** 3


# Cria um DataFrame com duas colunas:
# X → valores de x
# Y → valores de y
#
# Exemplo:
# X   Y
# 1   1
# 2   8
# 3   27
# ...
df_curva = pd.DataFrame({
    'X': x_curva,
    'Y': y_curva
})


# ============================================================
# CALCULANDO PEARSON
# ============================================================

# Calcula a correlação de Pearson entre X e Y.
#
# Pearson procura medir principalmente a relação LINEAR.
#
# method='pearson'
# → seleciona o método de Pearson.
pearson = df_curva['X'].corr(
    df_curva['Y'],
    method='pearson'
)


# ============================================================
# CALCULANDO SPEARMAN
# ============================================================

# Calcula a correlação de Spearman entre X e Y.
#
# Spearman trabalha com os RANKINGS (ordem) dos valores.
#
# Ele verifica se existe uma relação MONOTÔNICA:
# se X aumenta, Y tende a aumentar também.
#
# method='spearman'
# → seleciona o método de Spearman.
spearman = df_curva['X'].corr(
    df_curva['Y'],
    method='spearman'
)


# ============================================================
# MOSTRANDO OS RESULTADOS
# ============================================================

# Mostra o valor de Pearson.
#
# :.3f
# → mostra o número com 3 casas decimais.
print(f"Força segundo PEARSON (A Reta): {pearson:.3f}")


# Mostra o valor de Spearman.
#
# :.3f
# → 3 casas decimais.
print(f"Força segundo SPEARMAN (O Ranking): {spearman:.3f}")


# ============================================================
# CRIANDO O GRÁFICO
# ============================================================

# Define o tamanho da figura:
# largura = 6
# altura = 4
plt.figure(figsize=(6,4))


# Cria um gráfico de linha.
#
# x=x_curva → eixo X
# y=y_curva → eixo Y
#
# marker='o'
# → coloca uma bolinha em cada ponto.
#
# color='purple'
# → define a cor roxa.
sns.lineplot(
    x=x_curva,
    y=y_curva,
    marker='o',
    color='purple'
)


# Define o título do gráfico.
plt.title('A Cegueira do Pearson em Curvas Exponenciais')


# Exibe o gráfico.
plt.show()
```

```python
# [CÉLULA 5] SETUP DA OFICINA PBL (CALIFORNIA HOUSING)

# Importa o conjunto de dados California Housing
# diretamente da biblioteca Scikit-learn.
from sklearn.datasets import fetch_california_housing


# Baixa/carrega os dados do California Housing.
data = fetch_california_housing()


# Cria um DataFrame do Pandas.
#
# data.data → contém os valores das características
# data.feature_names → contém os nomes originais das colunas
#
# Cada linha representa uma região da Califórnia.
df = pd.DataFrame(
    data.data,
    columns=data.feature_names
)


# Cria a coluna 'Preco' com o preço das casas.
#
# data.target contém os preços em uma escala reduzida.
# Multiplicamos por 100000 para transformar os valores
# para uma escala aproximada de preço real em dólares.
df['Preco'] = data.target * 100000


# Renomeia as colunas originais para nomes em português.
#
# Exemplo:
# MedInc → Renda_Bairro
# HouseAge → Idade_Casa
# AveRooms → Qtd_Quartos
df = df.rename(columns={
    'MedInc': 'Renda_Bairro',       # Renda média do bairro
    'HouseAge': 'Idade_Casa',       # Idade média das casas
    'AveRooms': 'Qtd_Quartos',      # Quantidade média de quartos
    'AveBedrms': 'Qtd_Suites',      # Quantidade média de quartos/dormitórios
    'Population': 'Populacao',      # População da região
    'AveOccup': 'Ocupacao_Media',   # Ocupação média das casas
    'Latitude': 'Latitude',         # Latitude da região
    'Longitude': 'Longitude'        # Longitude da região
})


# Exibe as 5 primeiras linhas do DataFrame.
#
# head() → mostra as primeiras linhas para verificar
# se os dados foram carregados e organizados corretamente.
df.head()
```

### [CÉLULA 6] DESAFIO 1: A Matriz de Correlação
**O Problema (O Pedido do CEO):**
Qual característica estrutural ou social do imóvel tem o MAIOR poder estatístico para influenciar/ditar o seu Preço de Venda?

**Missão do Código:** Use a função `.corr()` do Pandas para gerar a matriz completa. Salve na variável `matriz_correlacao`. Imprima na tela apenas a coluna `Preco`, ordenando os valores do maior para o menor.

**Interpretação:** Adicione um comentário detalhando qual variável possui maior correlação com o preço e qual a relevância da 'Idade da Casa'.

```python
# [CÉLULA 7] RESOLUÇÃO - DESAFIO 1

# Calcula a matriz de correlação entre todas as colunas numéricas
# do DataFrame.
#
# A correlação varia de -1 até +1:
# +1 → correlação positiva perfeita
#  0 → ausência de correlação linear
# -1 → correlação negativa perfeita
matriz_correlacao = df.corr()


# Seleciona apenas a coluna 'Preco' da matriz de correlação.
#
# Isso permite descobrir a correlação de cada variável
# com o preço das casas.
#
# sort_values(ascending=False)
# → organiza os resultados do maior para o menor.
print(matriz_correlacao['Preco'].sort_values(ascending=False))


# ===== INTERPRETAÇÃO DOS RESULTADOS =====

# A 'Renda_Bairro' apresenta a maior correlação positiva
# com o 'Preco', aproximadamente 0.68.
#
# Isso indica uma relação linear positiva relativamente forte:
# conforme a renda média do bairro aumenta, o preço das casas
# tende a aumentar também.


# A 'Idade_Casa' possui uma correlação positiva bem menor,
# aproximadamente 0.10.
#
# Isso indica uma relação linear muito fraca entre a idade
# das casas e o preço.


# De forma geral, os resultados indicam que, neste conjunto
# de dados, a 'Renda_Bairro' possui uma relação linear
# mais forte com o preço do que a 'Idade_Casa'.
#
# IMPORTANTE:
# Correlação não significa causalidade.
# Uma correlação maior não prova que uma variável
# necessariamente causa a outra.
```

### [CÉLULA 12] DESAFIO 4 (BOSS): A Filtragem do Ruído
Sistemas limitados geram lixo estatístico, tracionando o vetor matemático e diluindo a força da correlação verdadeira.

**Missão do Código:** Estabeleça um filtro lógico no Pandas para compilar uma nova base `df_limpo` omitindo os registros de meio milhão ou mais. Realize a reextração de Pearson entre Renda e Preço.

**Interpretação:** Descreva qualitativamente o que ocorreu com a integridade matemática da correlação.

```python
# [CÉLULA 13] RESOLUÇÃO - DESAFIO 4


# ============================================================
# 1. FILTRO BOOLEANO DE SANEAMENTO
# ============================================================

# Remove do DataFrame os imóveis que possuem preço
# igual ou superior a US$ 500.000.
#
# df['Preco'] < 500000
# → cria uma condição booleana:
#    True  → preço menor que 500.000
#    False → preço maior ou igual a 500.000
#
# Apenas as linhas que possuem True são mantidas.
#
# O resultado é armazenado em um novo DataFrame chamado
# 'df_limpo', sem alterar o DataFrame original.
df_limpo = df[df['Preco'] < 500000]


# ============================================================
# 2. RECÁLCULO DA MATRIZ DE CORRELAÇÃO
# ============================================================

# Calcula novamente a correlação entre todas as variáveis,
# mas agora utilizando apenas os dados filtrados.
matriz_limpa = df_limpo.corr()


# Seleciona especificamente a correlação entre:
# 'Renda_Bairro' e 'Preco'.
#
# .loc[linha, coluna]
# → permite acessar um valor específico da matriz.
#
# O resultado é armazenado em 'coeficiente_puro'.
coeficiente_puro = matriz_limpa.loc['Renda_Bairro', 'Preco']


# ============================================================
# COMPARANDO OS RESULTADOS
# ============================================================

# Mostra a correlação ORIGINAL entre Renda_Bairro e Preco.
#
# matriz_correlacao
# → matriz calculada antes da limpeza.
#
# .loc['Renda_Bairro', 'Preco']
# → pega somente a correlação entre essas duas variáveis.
#
# :.4f
# → mostra o resultado com 4 casas decimais.
print(
    f"Força de Pearson anterior: "
    f"{matriz_correlacao.loc['Renda_Bairro', 'Preco']:.4f}"
)


# Mostra a correlação DEPOIS da limpeza.
#
# Agora utilizamos a matriz calculada com os dados filtrados.
print(
    f"Força de Pearson com o ruído cortado: "
    f"{coeficiente_puro:.4f}"
)


# ===== INTERPRETAÇÃO DOS RESULTADOS =====

# Depois da remoção dos valores no teto de US$ 500.000,
# a correlação entre Renda_Bairro e Preco aumenta
# de aproximadamente 0.68 para aproximadamente 0.70.


# Isso mostra que valores anômalos ou dados censurados
# podem influenciar os resultados de uma análise estatística.


# Por isso, antes de utilizar os dados em modelos de
# Machine Learning/Inteligência Artificial, é importante
# realizar uma etapa de limpeza e preparação dos dados.
#
# Essa etapa é chamada de pré-processamento ou
# higienização dos dados.
```
